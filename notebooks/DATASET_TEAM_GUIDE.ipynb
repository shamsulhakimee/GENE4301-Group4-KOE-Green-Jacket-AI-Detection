## 1. 🤔 What is a Dataset and Why Does It Matter?

An AI object detection model **learns by example** — you show it thousands of images, and it figures out patterns.  
For our project, we need to teach the AI:
- ✅ **What the KOE Green Jacket LOOKS LIKE** → these are called **Positive Samples**
- ❌ **What it does NOT look like** (so it doesn't make mistakes) → **Negative Samples**

Think of it like teaching a child to recognise a mango:
- You show them many mangoes at different angles, sizes, lighting → they learn 'mango'
- You also show them oranges, papayas, guavas → they learn 'not mango'

Without enough variety, the AI gets **confused**. Without negative samples, it might call every green thing a KOE jacket.

## 6. 🏷️ How to Label Images in Roboflow (Step-by-Step Beginner Guide)

### What is Roboflow?
Roboflow is a free web platform where you:
1. Upload your images
2. Draw boxes around the jackets (this is called 'annotation' or 'labelling')
3. Apply data augmentation (automatic image variations)
4. Export the dataset in YOLO format (ready for AI training)

### Step-by-Step: Setting Up the Project (Member 4 does this with PM Lead)
```
1. Go to: https://roboflow.com → Sign up with your email
2. Create a new Project:
   - Project Name: KOE-Green-Jacket-PPE
   - Annotation Group: Object Detection
   - Add Class: KOE_Green_Jacket (use exactly this name, no spaces)
3. PM Lead will share the workspace with all team members
```

### Step-by-Step: Uploading Images
```
1. Click 'Upload' in the Roboflow project
2. Drag and drop your image folder (can do batch of 50-100 at once)
3. Click 'Finish Uploading'
4. Images go to the 'Annotate' queue — ready for labelling
```

### Step-by-Step: Drawing Bounding Boxes (The Most Important Skill)

A **bounding box** is a rectangle you draw around the KOE jacket in each image.

```
1. Click on any image in the Annotate queue
2. Select the 'Bounding Box' tool (rectangle icon, or press 'B')
3. Click and drag to draw a rectangle around the KOE Green Jacket
   - Include the FULL jacket (shoulder to waist)
   - Keep the box TIGHT — don't include lots of empty space
   - Select class: KOE_Green_Jacket
4. If multiple people wear jackets in one image → draw a box for EACH person
5. Press the right arrow key → go to next image
6. Press 'Save' or it saves automatically
```

### Bounding Box Rules (CRITICAL — Read Carefully):

| Rule | Correct | Wrong |
| :--- | :--- | :--- |
| **Tightness** | Box hugs the jacket edges | Box has lots of background inside |
| **Completeness** | Full jacket from shoulder to waist | Half the jacket cut off by box edge |
| **Occlusion** | Still draw box even if partially hidden | Skipping a partially visible jacket |
| **Multiple people** | One box per person wearing jacket | One big box around all people |
| **Negatives** | No box drawn (image has no KOE jacket) | Drawing a box on a green t-shirt |

### What to DO with negative sample images:
> Upload them into Roboflow BUT **do not draw any boxes on them**.  
> Leave them empty. The AI will learn that images with no boxes = no KOE jacket present.

## 7. 📂 Understanding the Dataset File Structure (for context only)

After you export from Roboflow, the dataset will look like this folder structure:

```
dataset/
├── data.yaml               ← Tells the AI: 1 class = 'KOE_Green_Jacket'
├── train/
│   ├── images/
│   │   ├── photo_001.jpg   ← Training images (70% of total)
│   │   ├── photo_002.jpg
│   │   └── ...
│   └── labels/
│       ├── photo_001.txt   ← Bounding box coordinates for photo_001.jpg
│       ├── photo_002.txt
│       └── ...
├── valid/                  ← Validation images (20% of total)
│   ├── images/
│   └── labels/
└── test/                   ← Test images (10% of total — NEVER used during training)
    ├── images/
    └── labels/
```

### What's inside a label (.txt) file?
Each label file has one line per bounding box:
```
0  0.512  0.438  0.245  0.612
│    │      │      │      │
│    │      │      │      └── Box height (fraction of image height)
│    │      │      └───────── Box width (fraction of image width)
│    │      └──────────────── Box center Y (fraction of image height)
│    └─────────────────────── Box center X (fraction of image width)
└──────────────────────────── Class ID (0 = KOE_Green_Jacket, our only class)
```
> **Don't worry about this format** — Roboflow generates these label files automatically when you export.
> You just need to draw the boxes; Roboflow handles the rest.

## 8. ✂️ Train / Validation / Test Split Explained Simply

When we have 640 images, we split them into 3 groups:

```
TOTAL 640 IMAGES
│
├── 70% = 448 images → TRAIN set
│   Used to teach the AI. The AI sees these images during learning.
│
├── 20% = 128 images → VALIDATION set
│   Used to check if the AI is learning correctly while training.
│   The AI does NOT learn from these — they are used to measure progress.
│
└── 10% = 64 images → TEST set
    Hidden from the AI completely until the very end.
    Used ONCE to give a final, honest grade on performance.
```

**Roboflow sets this split automatically.** You just select 70/20/10 in the settings when generating the dataset version.

### Why do we need validation and test?
Imagine if you studied for an exam using the same questions that will appear in the exam — you'd score 100% but learn nothing real.  
Validation/test sets ensure the AI is **truly learning**, not just memorising the training images.

## 9. ✨ Data Augmentation — Making 640 Images into 2000+

**Augmentation** means automatically creating variations of your photos to give the AI more to learn from.
Roboflow does this automatically when you generate your dataset version.

### What augmentations we will apply (set in Roboflow):

| Augmentation | What It Does | Why Useful |
| :--- | :--- | :--- |
| **Horizontal Flip** | Mirror image left-right | Students enter from both sides |
| **Rotation ±12°** | Slightly tilt the image | Camera might not be perfectly straight |
| **Brightness ±20%** | Lighter and darker versions | Workshop lighting varies by time of day |
| **Saturation ±15%** | More or less colourful | Different camera colour profiles |
| **Mosaic** | Combines 4 images into 1 grid | Helps with detecting small and overlapping objects |
| **Cutout** | Random black squares on image | Teaches AI to handle partial occlusion |

> **How to apply in Roboflow:**
> 1. Go to 'Generate' tab in your project
> 2. Click 'Augmentation'
> 3. Enable the augmentations listed above with those values
> 4. Set 'Output per image: 3' (this triples your dataset size)
> 5. Click 'Generate Dataset'

## 11. 🚨 Common Annotation Mistakes (Review Weekly)

| Mistake | Problem It Causes | Fix |
| :--- | :--- | :--- |
| Box too loose (lots of background) | AI learns background, not jacket | Retighten box to jacket edges |
| Missing a person wearing jacket | AI misses those detections | Re-annotate; add the missing box |
| Box on a green shirt (wrong!) | AI confuses shirts with jackets | Delete that box immediately |
| Inconsistent labels (box size varies wildly) | Unstable detection | Follow a consistent annotation style |
| Only annotated from one angle | AI fails on other angles | Must have variety per guidelines above |
